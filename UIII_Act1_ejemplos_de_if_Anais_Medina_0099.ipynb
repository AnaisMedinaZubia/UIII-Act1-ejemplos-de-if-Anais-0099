#### Ejemplo de if

```python
# 1. Ejemplo de if simple
# Se utiliza para ejecutar un bloque de código solo si la condición se cumple
nota = 8  # Asigna el valor 8 a la variable nota [9]
if nota >= 6:  # Evalúa si la nota es mayor o igual a 6 (operador relacional) [10]
    print("El estudiante ha aprobado.")  # Se ejecuta solo si la condición es verdadera [1]
```

```python
# 1. Ejemplo de if simple
# Se utiliza para ejecutar un bloque de código solo si la condición se cumple
nota = 2  # Asigna el valor 8 a la variable nota [9]
if nota >= 6:  # Evalúa si la nota es mayor o igual a 6 (operador relacional) [10]
    print("El estudiante ha aprobado.")  # Se ejecuta solo si la condición es verdadera [1]
```

```python
# 2. Ejemplo de if else
# Permite manejar dos alternativas: una si la condición es verdadera y otra si es falsa
edad = 17  # Define la edad de la persona [11]
tiene_permiso = True  # Variable booleana de estado [13]

if edad >= 18 or tiene_permiso:  # Usa el operador relacional (>=) y el lógico (or) [7, 8]
    print("Acceso concedido.")  # Se ejecuta si la edad es >= 18 O si tiene permiso [8]
else:  # Rama que se ejecuta si ninguna de las condiciones anteriores es verdadera [1]
    print("Acceso denegado.")  # Mensaje para el caso falso
```

```python
# 3. Ejemplo de if elif else con Operadores Aritméticos
# Esta estructura permite evaluar múltiples condiciones de forma secuencial
numero = 10  # Asigna un valor numérico [9]
resultado = numero % 2  # Calcula el resto de la división usando el operador módulo [6]

if numero > 0 and resultado == 0:  # Verifica si es positivo Y si el resto es 0 (par) [8, 15]
    print("Es un número positivo y par.")  # Ejecuta si ambas condiciones lógicas se cumplen
elif numero > 0 and resultado != 0:  # Si falla la primera, verifica si es positivo e impar [8]
    print("Es un número positivo e impar.")  # Acción para la segunda condición
elif numero < 0:  # Evalúa si el número es menor que cero (negativo) [16]
    print("Es un número negativo.")  # Acción si la comparación es verdadera
else:  # Caso por defecto si no es positivo ni negativo [16]
    print("El número es cero.")  # Impresión para el valor neutro
```

```python
# 4. Ejemplo complejo: Año Bisiesto
# Este ejemplo integra cálculos aritméticos, comparaciones múltiples y operadores lógicos anidados
anio = 2024  # Asignación del año a evaluar [9]

# Verifica si es divisible por 4 y no por 100, O si es divisible por 400 [17, 18]
if (anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
    print(f"{anio} es un año bisiesto.")  # Mensaje si se cumple la lógica de bisiesto [17]
else:  # Caso contrario [1]
    print(f"{anio} no es un año bisiesto.")  # Mensaje para años comunes [17]
```
